```python
#opencv
import cv2

#matplotlib
import matplotlib.pyplot as plt

#pillow
from PIL import Image
```

```python
#reading image
img_one = cv2.imread(r'img/seminar.jpg')
print('shape of image one: ', img_one.shape)

print(img_one)
print(img_one.dtype)

plt.imshow(img_one)
plt.title('Image One')
plt.axis('off')
```

```python
img = cv2.imread('img/seminar.jpg',cv2.IMREAD_UNCHANGED)
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6), dpi=100)
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
```

```python
img_two_cv = cv2.imread(r'images/colors.png')
img_two_pil = Image.open(r'images/colors.png')

plt.subplot(1, 2, 1)
plt.imshow(img_two_cv)
plt.title('img_two_cv')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_two_pil)
plt.title('img_two_pil')
plt.axis('off')

# PIL: RGB --> CV: BGR
```

```python
# COLOR SPACE CONVERSION
img_two_cv_rgb = cv2.cvtColor(img_two_cv, cv2.COLOR_RGB2BGR)
img_two_pil = Image.open(r'images/colors.png')

plt.subplot(1, 2, 1)
plt.imshow(img_two_cv_rgb)
plt.title('img_two_cv_rgb')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_two_pil)
plt.title('img_two_pil')
plt.axis('off')

# RGB --> GBR (NO DIFFERENCE TO DISPAY)
```

```python
import cv2
from matplotlib import pyplot as plt

# Read and convert
img = cv2.imread(r'images/colors.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# Display using matplotlib
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.show()

print('shape of gray image: ', gray.shape)
```
