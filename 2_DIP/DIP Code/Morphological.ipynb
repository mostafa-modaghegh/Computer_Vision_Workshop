```python
import cv2
import matplotlib.pyplot as plt 
import numpy as np
```

```python
def show_img_plt(img, c_map ='gray', fig_hight=8, fig_width=6, vmin=0, vmax=255):
    plt.figure(figsize=(fig_hight, fig_width), dpi=100)
    plt.imshow(img, cmap=c_map, vmin=vmin, vmax=vmax)
```

```python
def show_mult_img(rows, columns, img_names, vmin=0, vmax=255):
    fig = plt.figure(figsize=(15, 17), dpi=100)
    for i in range(len(img_names)):
        fig.add_subplot(rows, columns, i+1)
        plt.imshow(img_names[i], cmap='gray', vmin=vmin, vmax=vmax)
        plt.axis('off')
        plt.title('img_' + str(i))
```

## Morphological Ops. 

```python
gharjistan_img = cv2.imread(r'images/Presentation1.jpg', 0)
show_img_plt(gharjistan_img)
```

```python
kernel = np.ones((5, 5), np.uint8)
dilated_img = cv2.dilate(gharjistan_img, kernel, iterations=5)
show_mult_img(1, 2, (gharjistan_img, dilated_img))
```

```python
eroded_img = cv2.erode(gharjistan_img, kernel, iterations=5)
show_mult_img(1, 2, (gharjistan_img, eroded_img))
```

```python
show_mult_img(1, 3, (gharjistan_img, dilated_img, eroded_img))
```
